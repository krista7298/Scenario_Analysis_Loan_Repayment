```python

# Step 2: Loan Repayment Scenario Analysis

from datetime import datetime
import pandas as pd
from IPython.display import display, Markdown

display(Markdown("### Loan Repayment Scenario Analysis"))
display(Markdown(f"Date: {datetime.now().strftime('%Y-%m-%d')}"))
display(Markdown("Student ID: KriAgu7298"))

# Function to calculate monthly loan payments
def calculate_payment(principal, annual_rate, years):
    r = annual_rate / 100 / 12
    n = years * 12
    monthly_payment = principal * r * (1 + r)**n / ((1 + r)**n - 1)
    return monthly_payment

# Loan amount
principal = 300000

# Three required scenarios
scenarios = [
    ("Scenario 1", 3, 15),
    ("Scenario 2", 5, 30),
    ("Scenario 3", 8, 40)
]

# Calculate monthly and total repayments
results = []

for name, rate, years in scenarios:
    monthly = calculate_payment(principal, rate, years)
    total = monthly * years * 12

    results.append({
        "Scenario": name,
        "Interest Rate (%)": rate,
        "Loan Term (Years)": years,
        "Monthly Payment ($)": round(monthly, 2),
        "Total Repayment ($)": round(total, 2)
    })

# Display results
loan_results = pd.DataFrame(results)
display(loan_results)
```

```python

# Step 3: Visualize Loan Repayment Scenarios

import matplotlib.pyplot as plt
from datetime import datetime
from IPython.display import display, Markdown

display(Markdown("### Loan Repayment Comparison"))
display(Markdown(f"Date: {datetime.now().strftime('%Y-%m-%d')}"))
display(Markdown("Student ID: KriAgu7298"))

# Create bar chart using Step 2 results
plt.figure(figsize=(9, 6))

bars = plt.bar(
    loan_results["Scenario"],
    loan_results["Total Repayment ($)"],
    color=["steelblue", "orange", "seagreen"]
)

# Add repayment amounts above bars
for bar in bars:
    height = bar.get_height()
    plt.text(
        bar.get_x() + bar.get_width() / 2,
        height,
        f"${height:,.2f}",
        ha="center",
        va="bottom"
    )

plt.title("Total Loan Repayment by Scenario")
plt.xlabel("Loan Scenario")
plt.ylabel("Total Repayment ($)")
plt.ylim(0, loan_results["Total Repayment ($)"].max() * 1.15)
plt.grid(axis="y", alpha=0.3)
plt.tight_layout()
plt.show()
```

```python

# Step 4: Sensitivity Analysis

import pandas as pd
import matplotlib.pyplot as plt
from datetime import datetime
from IPython.display import display, Markdown

display(Markdown("### Loan Repayment Sensitivity Analysis"))
display(Markdown(f"Date: {datetime.now().strftime('%Y-%m-%d')}"))
display(Markdown("Student ID: KriAgu7298"))

# Keep loan amount and term the same
principal = 300000
years = 30

# Test different interest rates
interest_rates = [3, 5, 8]

sensitivity_results = []

for rate in interest_rates:
    monthly = calculate_payment(principal, rate, years)
    total = monthly * years * 12

    sensitivity_results.append({
        "Interest Rate (%)": rate,
        "Monthly Payment ($)": round(monthly, 2),
        "Total Repayment ($)": round(total, 2)
    })

# Display results
sensitivity_df = pd.DataFrame(sensitivity_results)
display(sensitivity_df)

# Create sensitivity analysis graph
plt.figure(figsize=(9, 6))

bars = plt.bar(
    sensitivity_df["Interest Rate (%)"].astype(str) + "%",
    sensitivity_df["Total Repayment ($)"],
    color=["steelblue", "orange", "seagreen"]
)

for bar in bars:
    height = bar.get_height()
    plt.text(
        bar.get_x() + bar.get_width() / 2,
        height,
        f"${height:,.2f}",
        ha="center",
        va="bottom"
    )

plt.title("Sensitivity Analysis: Interest Rate vs. Total Repayment")
plt.xlabel("Annual Interest Rate")
plt.ylabel("Total Repayment ($)")
plt.ylim(0, sensitivity_df["Total Repayment ($)"].max() * 1.15)
plt.grid(axis="y", alpha=0.3)
plt.tight_layout()
plt.show()
```

### 1. What differences were observed between the scenarios?

The three scenarios showed that interest rates and loan lengths make a big difference in how much a loan costs. Scenario 1 had the lowest total repayment because it had a 3% interest rate and a 15-year term. Scenario 2 had a higher total repayment with a 5% interest rate over 30 years. Scenario 3 was the most expensive because it had an 8% interest rate and a 40-year term. Even though longer loans can have lower monthly payments, they can cost much more overall because interest builds up over time.

### 2. How did changes in key variables affect the outcomes in the sensitivity analysis?

For the sensitivity analysis, I kept the loan amount at $300,000 and the loan term at 30 years while changing the interest rate between 3%, 5%, and 8%. The results showed that higher interest rates increased both the monthly payment and the total repayment amount. This helped me see how much the interest rate affects the overall cost of borrowing money.
